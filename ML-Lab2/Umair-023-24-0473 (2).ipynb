```python
#Task 2
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5, 6, 7]
temp_city_a = [22, 24, 23, 26, 28, 27, 25]
temp_city_b = [18, 19, 21, 20, 22, 23, 21]
temp_city_c = [15, 18, 23, 25, 33, 38, 25]
plt.plot(days, temp_city_a, label='City A', color='tab:blue', marker='o')
plt.plot(days, temp_city_b, label='City B', color='tab:orange', marker='s')
plt.plot(days, temp_city_c, label='City C', color='tab:green', marker='p')
plt.title('Weekly Temperature Comparison')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.show()
```

```python
#Task 3
import matplotlib.pyplot as plt
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_scores = [52, 58, 63, 70, 75, 82, 88, 94]


color=[]
for score in exam_scores:
    if score < 70:
        color.append('red')
    else:
        color.append('green')
plt.scatter(study_hours, exam_scores, color=color, s=80, edgecolor='black')        
plt.title('Study Hours vs. Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.show()    
```

```python
#Task 4
import matplotlib.pyplot as plt

products = ['A', 'B', 'C', 'D']
q1_sales = [120, 95, 140, 80]
q2_sales = [130, 100, 120, 90]

import numpy as np
x = np.arange(len(products))
width = 0.35

plt.bar(x - width/2, q1_sales, width, label='Q1')
plt.bar(x + width/2, q2_sales, width, label='Q2')
plt.xticks(x, products)
plt.title('Quarterly Sales by Product')
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.legend()
plt.show()
```

```python
#Task 5 #Box Plot
import matplotlib.pyplot as plt
section_a = [65, 70, 72, 75, 78, 80, 82, 85, 88, 95]
section_b = [55, 60, 62, 65, 68, 70, 72, 74, 76, 98]

plt.boxplot([section_a, section_b], labels=['Section A', 'Section B'])
plt.title('Exam Score Distribution by Section')
plt.ylabel('Score')
plt.show()
```

```python
#Task 6
import matplotlib.pyplot as plt
categories = ['Housing', 'Food', 'Transport', 'Entertainment', 'Savings']
budget = [35, 20, 15, 10, 20]
plt.pie(budget, labels=categories, autopct='%1.1f%%', startangle=90,explode=(0., 0, 0, 0, 0.05),colors=['skyblue', 'lightgreen', 'orange', 'pink', 'purple'])
plt.title('Monthly Budget Allocation')
plt.show()
```

```python
#Task 7

import matplotlib.pyplot as plt
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].plot(days, temp_city_a, color='tab:blue')
axes[0, 0].set_title('Line: Temperature')
axes[0, 1].scatter(study_hours, exam_scores, color='tab:green')
axes[0, 1].set_title('Scatter: Study vs. Score')
axes[1, 0].bar(products, q1_sales, color='tab:orange')
axes[1, 0].set_title('Bar: Q1 Sales')
axes[1, 1].boxplot([section_a, section_b], labels=['A', 'B'])
axes[1, 1].set_title('Box: Exam Scores')

fig.suptitle('Mini Dashboard', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
#Task 8
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
print(df.head())
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
 
axes[0, 0].hist(df['sepal length (cm)'], bins=15, color='tab:blue')
axes[0, 0].set_title('Sepal Length Distribution')
 
colors = df['species'].map({'setosa': 'tab:blue', 'versicolor': 'tab:orange', 'virginica': 'tab:green'})
axes[0, 1].scatter(df['petal length (cm)'], df['petal width (cm)'], c=colors)
axes[0, 1].set_title('Petal Length vs. Width by Species')
 
df.boxplot(column='sepal width (cm)', by='species', ax=axes[0, 2])
axes[0, 2].set_title('Sepal Width by Species')
 
df['species'].value_counts().plot(kind='bar', ax=axes[1, 0], color='tab:purple')
axes[1, 0].set_title('Samples per Species')
 
df['species'].value_counts().plot(kind='pie', ax=axes[1, 1], autopct='%1.0f%%')
axes[1, 1].set_title('Species Proportions')
 
axes[1, 2].plot(df.index, df['sepal length (cm)'], color='tab:red', alpha=0.6)
axes[1, 2].set_title('Sepal Length Across Samples')
 
fig.suptitle('Iris Dataset — Visual EDA Report', fontsize=15)
plt.tight_layout()
plt.show()
```

```python
# TASK 8 #
HISTOGRAM: Most flowers have a sepal length between 5 and 7 cm.

SCATTER PLOT: The species have different petal lengths and widths.

BOX PLOT: Sepal width is different for each species.

BAR CHART: Each species has 50 samples.

PIE CHART: Each species has about 33% of the samples.

LINE CHART: Sepal length changes across the samples.
```

```python
# 1.  #Line Graph #
import matplotlib.pyplot as plt

days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
steps = [4500, 5200, 6100, 5800, 7000, 8500, 7600]
plt.plot(days, steps, marker='o')
plt.title('Weekly Daily Steps')
plt.xlabel('Days')
plt.ylabel('Number of Steps')
plt.grid(True)
plt.show()
```

```python
# 2. #Scatter Graph
import matplotlib.pyplot as plt
sleep = [4, 5, 5.5, 6, 6.5, 7, 7.5, 8, 8.5, 9]
productivity = [3, 4, 4.5, 5, 5.5, 6, 6.5, 7, 8, 9]
colors = ['red' if score < 5 else 'blue' for score in productivity]
plt.scatter(sleep, productivity, c=colors, s=80)
plt.title('Hours of Sleep vs Productivity Score')
plt.xlabel('Hours of Sleep')
plt.ylabel('Productivity Score')
plt.show()
```

```python
# 3. #Bar Graph
import matplotlib.pyplot as plt

products = ['Product A','Product B','Product C']
q1_sales = [40, 55, 35]
q2_sales = [50, 65, 45]

import numpy as np
x = np.arange(len(products))
width = 0.35

plt.bar(x - width/2, q1_sales, width, label='Q1')
plt.bar(x + width/2, q2_sales, width, label='Q2')
plt.xticks(x, products)
plt.title('Product Sales Across Two Quarters')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
# 4. #Box Plot
import matplotlib.pyplot as plt
group_a = [10, 12, 13, 14, 15, 16, 17]
group_b = [5, 8, 10, 12, 18, 22, 25]
group_c = [20, 21, 22, 23, 24, 25, 26]
plt.boxplot([group_a, group_b, group_a], labels=['Group A','Group B','Group C'])
plt.title('Camparison of Three Groups')
plt.ylabel('Values')
plt.show()
print('Group B has the widest spread.')
```

```python
# 5. #Pie chart
import matplotlib.pyplot as plt
categories = ['Study', 'Sleep', 'Exercise', 'Entertainment', 'Other']
time = [30, 35, 10, 15, 10]
plt.pie(time, labels=categories, autopct='%1.1f%%', explode=[0.1, 0, 0, 0, 0], startangle=90)
plt.title('Weekly Time Allocation')
plt.show()
print('Total =', sum(time), '%')
```

```python
# 6. #Sub Plots
import matplotlib.pyplot as plt
import numpy as np
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Four Chart Types', fontsize=16)

#line
axes[0, 0].plot(days, steps, marker='o')
axes[0, 0].set_title('Weekly Daily Steps')
axes[0, 0].set_xlabel('Days')
axes[0, 0].set_ylabel('No. of Steps')

#Scatter
axes[0, 1].scatter(sleep, productivity, c=colors, s=80)
axes[0, 1].set_title('Hours of Sleep vs Productivity Score')
axes[0, 1].set_xlabel('Sleep Hours')
axes[0, 1].set_ylabel('Productivity Score')

#Bar
axes[1, 0].bar(x - width/2, q1_sales, width, label='Q1')
axes[1, 0].bar(x + width/2, q2_sales, width, label='Q2')
axes[1, 0].set_title('Product Sales Across Two Quarters')
axes[1, 0].set_xticks(x)
axes[1, 0].set_xlabel('Products')
axes[1, 0].set_ylabel('Sales')
axes[1, 0].legend()

#Box
axes[1, 1].boxplot([group_a, group_b, group_a], labels=['Group A','Group B','Group C'])
axes[1, 1].set_title('Camparison of Three Groups')
axes[1, 1].set_ylabel('Values')

plt.tight_layout()
plt.show()
```
